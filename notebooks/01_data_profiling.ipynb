```python
import pandas as pd
import glob

# 1. Load semua data mentah dari folder data/raw
raw_files = glob.glob('../data/raw/*.csv')
df = pd.concat([pd.read_csv(f) for f in raw_files], ignore_index=True)

# 2. Cek 5 data teratas
print("--- 5 Sample Data Mentah ---")
display(df.head())

# 3. Cek Jumlah Baris, Kolom, dan Tipe Data
print("\n--- Ringkasan Dataset ---")
df.info()

# 4. Cek Jumlah Missing Value (NULL) per Kolom
print("\n--- Jumlah Missing Value (NULL) ---")
print(df.isnull().sum())

# 5. Cek Total Duplikasi Data
print("\n--- Total Baris Duplikat ---")
print(f"Jumlah duplikat: {df.duplicated().sum()}")
```
